```python
from pathlib import Path
import json
import sys
import pandas as pd
project_root = Path.cwd().parents[1]
sys.path.insert(0, str(project_root))

from tools.functional_evidence import analyze_variant

OUTPUT_SUMMARY_PATH = project_root / "outputs/batch_20260722_110732/batch_summary.json"
OUTPUT_SUMMARY_PARENT = OUTPUT_SUMMARY_PATH.parent

failed_variants = []

for file in sorted(OUTPUT_SUMMARY_PARENT.glob("*.json")):
    print("=== FILE NAME ===")
    print(file.name)

    if file.name == "batch_summary.json":
        continue

    with open(file, "r", encoding="utf-8") as f:
        payload = json.load(f)

    if "results" not in  payload:
        continue

    if not payload.get("results") or not isinstance(payload.get("results"), list):
        continue

    results = payload.get("results")

    ps3_result = next(
        (result for result in results if result.get("criterion", "") == "PS3"),
        None
    )

    bs3_result = next(
        (result for result in results if result.get("criterion", "") == "BS3"),
        None
    )

    if bs3_result is None or ps3_result is None:
        continue
    
    if ps3_result.get("status") == "error" or bs3_result.get("status") == "error":
        failed_variants.append(payload.get("variant", "unknown"))

failed_variants










```
